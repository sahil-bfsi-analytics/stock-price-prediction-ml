```python
df_encode=pd.get_dummies(df,drop_first=True)
df_encode.head()
```

```python
bool_cols = df_encode.select_dtypes(include='bool').columns

df_encode[bool_cols] = df_encode[bool_cols].astype(int)
```

```python
# ==========================================
# Pearson Correlation Test
# ==========================================

from scipy.stats import pearsonr
import pandas as pd

# Target
target = 'Close'

# Features to check against target
selected_features = [
    'Open',
    'High',
    'Low',
    'Volume',
    'Daily Return',
    'MA_7',
    'MA_30',
    'Volatility_30D',
    'Market Cap (Current, INR Cr)',
    'is_imputed',
    'NSE Symbol_TATAMOTORS',
    'NSE Symbol_TATAPOWER',
    'NSE Symbol_TATASTEEL',
    'NSE Symbol_TATATECH',
    'NSE Symbol_TCS'
]

# Pearson correlation
correlations = {
    feature: pearsonr(
        df_encode[feature],
        df_encode[target]
    )[0]
    for feature in selected_features
}

correlation_df = pd.DataFrame(
    list(correlations.items()),
    columns=['Feature', 'Pearson Correlation']
)

correlation_df = correlation_df.sort_values(
    by='Pearson Correlation',
    ascending=False
)

correlation_df
```

```python
# ==========================================
# Chi-Square Test for Categorical Features
# ==========================================

from scipy.stats import chi2_contingency

alpha = 0.05

cat_features = [
    'NSE Symbol_TATAMOTORS',
    'NSE Symbol_TATAPOWER',
    'NSE Symbol_TATASTEEL',
    'NSE Symbol_TATATECH',
    'NSE Symbol_TCS'
]

# Convert continuous target into bins
df_encode['Close_bin'] = pd.qcut(
    df_encode['Close'],
    q=4,
    labels=False,
    duplicates='drop'
)

chi2_results = {}

for col in cat_features:

    contingency = pd.crosstab(
        df_encode[col],
        df_encode['Close_bin']
    )

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)

    decision = (
        'Reject Null (Keep Feature)'
        if p_val < alpha
        else 'Accept Null (Drop Feature)'
    )

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T

chi2_df = chi2_df.sort_values(
    by='p_value'
)

chi2_df
```

```python
# Final selected features based on Pearson + Chi-Square results

selected_features = [
    'High',
    'Low',
    'Open',
    'MA_7',
    'MA_30',
    'NSE Symbol_TATASTEEL',
    'NSE Symbol_TATAPOWER',
    'Daily Return',
    'Market Cap (Current, INR Cr)'
]

# Create final dataframe
final_df = df_encode[selected_features + ['Close']].copy()

final_df
```

```python
final_df.head()
```

```python
import pickle
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

# X and y
X = final_df.drop(columns=['Close'])
y = final_df['Close']

# Train/Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# StandardScaler
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Linear Regression
model = LinearRegression()

model.fit(X_train_scaled, y_train)

# Save model
with open('model.pkl', 'wb') as file:
    pickle.dump(model, file)

# Save scaler
with open('scaler.pkl', 'wb') as file:
    pickle.dump(scaler, file)

print("model.pkl and scaler.pkl saved successfully!")
```
